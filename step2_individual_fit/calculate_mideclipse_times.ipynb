```python
import numpy as np
import astropy.units as u

from astropy.constants import c
from astropy.time import Time
from astropy.coordinates import (
    SkyCoord,
    get_body_barycentric,
    solar_system_ephemeris,
)


# ---------------------------------------------------------
# Inputs
# ---------------------------------------------------------

anchor_hjd_tdb = 2461034.435937993
period_days = 24.7372457

target = SkyCoord(
    ra="00h44m59.6827s",
    dec="-15d16m27.09s",
    frame="icrs",
)


# ---------------------------------------------------------
# HJD_TDB -> BJD_TDB
# ---------------------------------------------------------

def hjd_tdb_to_bjd_tdb(hjd_tdb, target):
    """
    Convert HJD_TDB to BJD_TDB.

    This uses:

        BJD_TDB - HJD_TDB
            = projection of the Sun's barycentric position
              along the target line of sight / c

    Parameters
    ----------
    hjd_tdb : float or array-like
        Heliocentric Julian Date in the TDB scale.

    target : astropy.coordinates.SkyCoord
        Target sky coordinates.

    Returns
    -------
    astropy.time.Time
        BJD_TDB values.
    """
    hjd_tdb = np.asarray(hjd_tdb, dtype=float)

    # The difference between HJD and BJD is only a few seconds,
    # so using HJD as the evaluation epoch is sufficient initially.
    t = Time(hjd_tdb, format="jd", scale="tdb")

    # Unit vector from the Solar System barycenter toward the target
    n_hat = target.icrs.cartesian.xyz.value
    n_hat = n_hat / np.linalg.norm(n_hat)

    with solar_system_ephemeris.set("builtin"):
        # Sun's position relative to the Solar System barycenter
        r_sun = get_body_barycentric("sun", t)

    # Shape is normally (3,) for one time or (3, N) for an array
    r_xyz = r_sun.xyz.to(u.m).value

    if r_xyz.ndim == 1:
        projected_distance = np.dot(r_xyz, n_hat)
    else:
        projected_distance = np.sum(
            r_xyz * n_hat[:, np.newaxis],
            axis=0
        )

    correction_days = (
        projected_distance * u.m / c
    ).to_value(u.day)

    bjd_tdb = Time(
        hjd_tdb + correction_days,
        format="jd",
        scale="tdb"
    )

    return bjd_tdb

def print_event(name, hjd, bjd):
    print(f"\n{name}")
    print("-" * len(name))
    print(f"HJD_TDB:  {hjd:.12f}")
    print(f"BJD_TDB:  {bjd.jd:.12f}")
    print(f"BMJD_TDB: {bjd.mjd:.12f}")
    print(f"UTC:      {bjd.utc.iso}")


```
